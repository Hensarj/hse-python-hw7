## Задание 1. Первые пять постов

Вывод ниже получен при реальном локальном запросе к JSONPlaceholder.


```python
import requests

# Получаем посты и преобразуем JSON в список словарей.
try:
    response = requests.get(
        "https://jsonplaceholder.typicode.com/posts", timeout=15
    )
    response.raise_for_status()
    posts = response.json()

    for post in posts[:5]:
        print("Заголовок:", post["title"])
        print("Текст:", post["body"])
        print()
except requests.exceptions.Timeout:
    print("Сервис не ответил вовремя. Попробуйте ещё раз.")
except requests.exceptions.RequestException:
    print("Не удалось получить посты. Проверьте подключение и сервис.")
except (ValueError, KeyError, TypeError):
    print("Не удалось прочитать данные постов.")
```

## Задание 2. Погода в выбранном городе

Запустите ячейку и введите город и свой действующий ключ OpenWeather в скрытом поле. Ключ не вставляйте в код и не отправляйте в чат. Нужны интернет и библиотека `requests`.

Успешный запрос погоды пока не выполнен: доступного ключа нет. Результат погоды не подставлен.


```python
import requests
from getpass import getpass

city = input("Введите название города: ").strip()

if not city:
    print("Название города не должно быть пустым.")
else:
    # Ввод скрыт: ключ не нужно записывать в код или публиковать.
    api_key = getpass("Ключ OpenWeather: ").strip()
    if not api_key:
        print("Для запроса нужен ваш ключ OpenWeather.")
    else:
        params = {
            "q": city,
            "appid": api_key,
            "units": "metric",  # Температура в градусах Цельсия.
            "lang": "ru",       # Описание погоды на русском.
        }
        try:
            response = requests.get(
                "https://api.openweathermap.org/data/2.5/weather",
                params=params,
                timeout=15,
            )
            if response.status_code == 401:
                print("Ключ OpenWeather неверен или ещё не активирован.")
            elif response.status_code == 404:
                print("Город не найден. Проверьте название.")
            else:
                response.raise_for_status()
                weather = response.json()
                print("Город:", city)
                print("Температура:", weather["main"]["temp"], "°C")
                print("Погода:", weather["weather"][0]["description"])
        except requests.exceptions.Timeout:
            print("Сервис не ответил вовремя. Попробуйте ещё раз.")
        except requests.exceptions.RequestException:
            # Не печатаем исключение: URL запроса может содержать ключ.
            print("Не удалось получить погоду. Проверьте сеть и сервис.")
        except (ValueError, KeyError, IndexError, TypeError):
            print("Не удалось прочитать данные о погоде.")
        finally:
            # Не оставляем ключ в этих переменных после выполнения.
            api_key = ""
            params.clear()
            response = None
```
